## 1. Prerequisites (R and Apollo)

Because Delphos acts as an intelligent assistant that sends specifications to R for estimation, you should ensure your R environment is fully configured *before* installing Delphos.

1. **Install R**: Download and install R from CRAN.
2. **Install RStudio** (Optional but highly recommended).
3. **Install Apollo**: Open RStudio and run `install.packages('apollo')`.

### ⚠️ Important step for Windows users: Add R to your PATH
On macOS and Linux, R is automatically added to your system `PATH`. However, on Windows, you must manually tell your command prompt where R is located so Delphos can find it.

1. Open your Windows Start Menu and type **"Environment Variables"** -> Click **"Edit the system environment variables"**.
2. Click the **"Environment Variables..."** button at the bottom.
3. Under the **"System variables"** list, find the variable named **`Path`**, select it, and click **Edit**.
4. Click **New** and paste the folder path where R's binaries are installed (e.g., `C:\Program Files\R\R-4.4.1\bin\x64`). Check your `C:\Program Files\R\` folder to find your exact version.
5. Click **OK** on all windows to save.
6. **Crucial:** You must completely close and restart your Anaconda Prompt, Terminal, and Jupyter Lab for the new PATH to take effect!

## 3. Verifying the Installation

Let's import the package and check that it successfully loads the agent and the default catalogue.

```python
import delphos as dp

print("Delphos successfully imported!")

# Load the default agent
agent = dp.load_agent()
print("Agent loaded successfully! Device:", agent.agent.device)

# Load the global catalogue
catalogue = dp.data.registry.load_global_catalogue()
print(f"Global catalogue loaded! It contains {catalogue.n_attributes} attributes and {catalogue.n_covariates} covariates.")
```
